```python
from tvb_multiscale.tvb_nest.config import Config
from tvb_multiscale.tvb_nest.nest_models.nest_ray import RayNESTServer

config = Config()
config.DEFAULT_LOCAL_NUM_THREADS = 4

RayNESTServer = ray.remote(RayNESTServer)
nest_server = RayNESTServer.options(name="nest_server", 
                                    num_cpus=config.DEFAULT_LOCAL_NUM_THREADS).remote(config=config)
# ray.kill(nest_server)
```

```python
# ray.get(nest_server.nest.remote("SetKernelStatus", {"local_num_threads": config.DEFAULT_LOCAL_NUM_THREADS}))
ray.get(nest_server.nest.remote("GetKernelStatus"))
```

```python
from tvb_multiscale.tvb_nest.nest_models.nest_ray import RayNESTClient

nest = RayNESTClient(nest_server)
```

```python
nest.GetKernelStatus("time")
```

```python
from tvb_multiscale.tvb_nest.nest_models.nest_ray import RayNodeCollection
from tvb_multiscale.tvb_nest.nest_models.node import NESTNodeCollection
n = NESTNodeCollection(nodes=RayNodeCollection(nest, ()))
```

```python
source = nest.Create("iaf_cond_alpha")
print(source.global_id)
```

```python
target = nest.Create("iaf_cond_alpha")
print(target.global_id)
```

```python
nest.Connect(source, target)
```

```python
conns = nest.GetConnections(source, target)
```

```python
[conns.source, conns.target, conns.weight, conns.delay, conns.receptor]
```

```python
nest.Prepare()
```

```python
import time

def run(t):
    print("NEST time = %g" % nest.GetKernelStatus("time"))
    print("simulating...")
    while nest.Run(t):
        print("...running...")
        time.sleep(1.0)
    print("...done!")
    print("NEST time = %g" % nest.GetKernelStatus("time"))

run(100000.0)
```
