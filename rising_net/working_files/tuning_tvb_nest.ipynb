```python
import os
from matplotlib import pyplot as plt
import numpy as np
from tvb_multiscale.core.utils.file_utils import load_pickled_dict

basepath = "/home/docker/packages/tvb-multiscale/rising_net/working_files/outputs/nest_tvb_"

mossy_rates =[]
for w_TVB_to_NEST in TUNED_VALUES_TVB_TO_NEST:
    filepath = os.path.join(basepath + "%g" % w_TVB_to_NEST, "res/rate_RMSE.pkl")
    res = load_pickled_dict(filepath)
    mossy_rates.append(res["mossy_rate"])

print(mossy_rates)
```

```python
plt.plot(TUNED_VALUES_TVB_TO_NEST, mossy_rates, "b")
plt.plot(TUNED_VALUES_TVB_TO_NEST, 3.9*np.ones((8,)), "r--")
plt.xlabel("w_TVB_to_NEST")
plt.ylabel("Mossy fibers rate (Hz)")
```
